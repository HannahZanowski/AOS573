````python
import numpy as np
import matplotlib.pyplot as plt
````

````python
nao_data=np.genfromtxt('nao_index.tim',skip_header=8,names=True)
len(nao_data)
````

````python
#Make the plot
#set up the plotting axis
fig,ax=plt.subplots()
#set the figure size in inches
fig.set_size_inches(12,3)
#Plot the index data
plt.plot(nao_data['INDEX'],color='r',lw=0.5)
yrs=np.arange(1950,2026,10) #create the years for the ticklabels
ax.set_xticks(np.arange(0,len(nao_data),120)) #Set the ticks to go by 120 months (10 years)
ax.set_xticklabels(yrs) #Set the tick labels to be in years
ax.set_xlim(0,len(nao_data)) #set the x-axis limits to span only the length of the data
plt.title('Monthly NAO Index') #plot title
plt.xlabel('Time (years)')
plt.ylabel ('Standardized NAO Index'); #Semi-colon here supresses the text output from the command
````

For this problem you will download time series of monthly sea ice area (units: million km$^{2}$) for the Arctic and Antarctic from November 1978 to August 2026

2. Use numpy's [genfromtxt](https://numpy.org/doc/stable/reference/generated/numpy.genfromtxt.html) function to read in the data (again, no using [pandas](https://pandas.pydata.org/) yet, but that doesn't mean you can't use pandas to check your work)

3. Make a 1x2 set of plots with a shared y-axis showing the annual cycle of a) Arctic and b) Antarctic sea ice extent (one region per subplot) that adhere to the following parameters:
> 1. Plot the sea ice area annual cycle (monthly output) every 5th year starting in 1980. That means your plots will have multiple lines representing Jan-Dec data for each of the years from 1980-2025. 
> 2. Color the lines using a matplotlib colormap, so that you can see a color gradient in your lines over time. I recommend using one of matplotlib's sequential colormaps for this.
> 3. Change the linewidth to be thinner than the default for each year plotted above
> 4. Plot the 2026 data for each region (it will only be through August 2026) as black dotted line with a linewidth of 2.5

````python
nh_sia=np.genfromtxt('n_icearea.mon.data',skip_header=1,skip_footer=11)
sh_sia=np.genfromtxt('s_icearea.mon.data',skip_header=1,skip_footer=11)

#mask the -99.99 value that represents missing data
nh_sia=np.ma.masked_where(nh_sia==-99.99,nh_sia)
sh_sia=np.ma.masked_where(sh_sia==-99.99,sh_sia)
````

````python
nh_5yrs=nh_sia[2::5]
sh_5yrs=sh_sia[2::5]
````

````python
#Get the colors from the colormap
cmap=plt.colormaps['RdPu'] #choose a colormap
colors=cmap(np.linspace(0, 1, len(nh_5yrs))) #map the correct number of colors that you'll need (needs to be the number of lines you are plotting)
#Month strings for plotting
months=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
#Subplot titles
tts=['Arctic Sea Ice Area','Antarctic Sea Ice Area']
#Tuples for legend location--anchor for lower left corner of legend in axes coords (x,y)
locs=[(0.01,0.01),(0.5,0.01)]
fig,ax=plt.subplots(1,2,figsize=(10,3),sharey=True)
for i in range(0,10):
    ax[0].plot(months,nh_5yrs[i,1:],label=nh_5yrs[i,0],color=colors[i],lw=1.0)
    ax[1].plot(months,sh_5yrs[i,1:],label=nh_5yrs[i,0],color=colors[i],lw=1.0)
ax[0].plot(months,nh_sia[-1,1:],label=nh_sia[-1,0],color='k',lw=2.5,ls=':')
ax[1].plot(months,sh_sia[-1,1:],label=sh_sia[-1,0],color='k',lw=2.5,ls=':')
#axes params and other plotting stuff
for i in range(0,2):
    ax[i].set_xlim(months[0],months[-1])
    ax[i].set_ylim(0,16)
    ax[i].legend(loc=locs[i],frameon=False,ncol=2,borderaxespad=0,handlelength=1,handletextpad=0.2,fontsize=9)
    ax[i].set_title(tts[i],fontsize=11)
    ax[i].set_xlabel('Month')
    ax[i].tick_params(labelsize=9)
ax[0].set_ylabel(r'Sea Ice Area ($\times$10$^{6}$ km$^{2}$)');
````
